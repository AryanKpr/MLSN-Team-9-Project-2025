````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
````

````python
path = "https://raw.githubusercontent.com/AryanKpr/MLSN-Team-9-Project-2025/main/data.csv"
data = pd.read_csv(path)
````

# Task
Create a step-by-step guide to generate a heatmap of the correlation matrix for the numerical columns in the DataFrame `df`.

## Select relevant columns

### Subtask:
Choose the numerical columns from the DataFrame that you want to include in the heatmap.


**Reasoning**:
Identify the numerical columns in the dataframe and store them in a list.



````python
numerical_cols = data.select_dtypes(include=np.number).columns.tolist()
print(numerical_cols)
````

## Calculate the correlation matrix

### Subtask:
Compute the pairwise correlation between the selected numerical columns.


**Reasoning**:
Select the numerical columns and compute the correlation matrix.



````python
numerical_data = data[numerical_cols]
correlation_matrix = numerical_data.corr()
````

## Generate the heatmap

### Subtask:
Create a heatmap using the correlation matrix to visualize the relationships between the numerical variables.


**Reasoning**:
Generate a heatmap of the correlation matrix to visualize the relationships between numerical variables.



````python
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
````

### Subtask:
Add the correlation values to the heatmap for better interpretation.


**Reasoning**:
Regenerate the heatmap with annotations and formatting for better interpretation.



````python
plt.figure(figsize=(14, 12))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Numerical Variables with Values')
plt.show()
````

## Customize the heatmap (optional)

### Subtask:
Adjust the colormap, title, and axis labels for better visualization.


**Reasoning**:
Adjust the colormap, title, and axis labels for better visualization of the heatmap.



````python
plt.figure(figsize=(14, 12))
sns.heatmap(correlation_matrix, annot=True, cmap='viridis', fmt=".2f")
plt.title('Correlation Matrix of Athlete Performance Metrics', fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
````

#Feature Engineering

Based on the correlation values these features have on the athelete's probability of having an injury next season, it would be wise to select features such as stress level (Corr. 0.54), reaction time (Corr. 0.47), and/or previous injury (Corr. 0.38). Some other features we can use are injuries per season (previous injury count/years active) and total workload (training hours * matches played).

````python
data['Total_Workload'] = data['Training_Hours_Per_Week'] * data['Matches_Played_Past_Season']
display(data[['Training_Hours_Per_Week', 'Matches_Played_Past_Season', 'Total_Workload']].head())
````

````python
data['Injuries_Per_Season'] = data['Previous_Injury_Count'] / data['Matches_Played_Past_Season']
display(data[['Previous_Injury_Count', 'Matches_Played_Past_Season', 'Injuries_Per_Season']].head())
````

````python
cols_to_use = data.drop(['Age','Height_cm', 'Weight_kg', 'Training_Hours_Per_Week', 'Matches_Played_Past_Season', 'Previous_Injury_Count',
                         'Knee_Strength_Score', 'Hamstring_Flexibility','Reaction_Time_ms', 'Balance_Test_Score', 'Agility_Score', 'Nutrition_Quality_Score', 'BMI','Position'],axis=1)
print(cols_to_use.columns)
````

````python

from sklearn.model_selection import train_test_split
X = cols_to_use.drop('Injury_Next_Season', axis=1)
y = cols_to_use['Injury_Next_Season']
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.7, test_size = 0.3)
````

````python
from sklearn.ensemble import RandomForestClassifier
model1 = RandomForestClassifier(random_state=20)
model1.fit(X_train, y_train)
````

````python
import joblib
filename = 'my_model.joblib'
joblib.dump(model1, filename)
from google.colab import files
files.download('my_model.joblib')
````

````python
from sklearn.linear_model import LogisticRegression
model2 = LogisticRegression(random_state=60,max_iter=200)
model2.fit(X_train, y_train)
````

````python
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

# Make predictions on the test set
y_pred = model1.predict(X_test)

# Calculate and print accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")

# Generate and plot the confusion matrix
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model1.classes_)
disp.plot(cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.show()
````

````python
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier(max_depth=3, random_state=60)
clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)

acc = accuracy_score(y_test, y_pred)
print('Accuracy (DecisionTreeClassifier):', acc)
````

**XG Boost Model**

````python
import xgboost as xgb
from sklearn.metrics import accuracy_score

xgb_model = xgb.XGBClassifier(random_state=42)

xgb_model.fit(X_train, y_train)

y_pred = xgb_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"XGBoost Accuracy: {accuracy:.4f}")
````
